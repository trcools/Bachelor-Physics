# $\color{orange}{\textit{Figuren bij Hoofdstuk 3: Paramagnetisme (situation for S=1/2)}}$

```
xval=np.linspace(-5.0,+5.0,num=300) # x = \frac { \mu \mathcal{B} } { k T}
xm1val=np.linspace(-6.0,+6.0,num=300) # 1 over x
```

```
# Figure 1: the probabilities p_+ and p_- as a function of x
#
fig1 = plt.figure()
fig1.subplots_adjust(bottom=0.2,left=0.2)
ax1  = fig1.add_subplot(111)
# compute the probabilities
p_plus= np.divide(np.exp(xval),2.*np.cosh(xval))
p_min = np.divide(np.exp(-xval),2.*np.cosh(xval))
#
#
ax1.plot(xval,p_plus,label=r'$p_{+}$',linestyle='-',linewidth=2,color='mediumslateblue')
ax1.plot(xval,p_min,label=r'$p_{-}$',linestyle='--',linewidth=2,color='mediumslateblue')
#
ax1.set_ylabel(r"Probabiliteit",fontsize=21,weight='bold')
ax1.set_xlabel(r" $\bf x = \frac { \mu \mathbb{B} } { k T}$",fontsize=21,weight='bold')
ax1.set_ylim(-0.05,1.05) ; ax1.set_xlim(0.,3.)
ax1.locator_params(axis='y',nbins=5) # control the number of y-axis ticks
ax1.legend(frameon=False,fontsize=22,ncol=1,loc='best')
fig1.savefig("fig3p1_n.eps")
fig1.savefig("fig3p1_n.pdf")
plt.show(fig1)
```

```
# Figure 2: the magnetisation density as a function of x
#
fig2 = plt.figure()
fig2.subplots_adjust(bottom=0.2,left=0.2)
ax2  = fig2.add_subplot(111)
# compute the magnetisation
PP= np.tanh(xval)
#
#
ax2.plot(xval,PP,linestyle='-',linewidth=2,color='mediumslateblue')
ax2.plot(xval,xval,linestyle='--',linewidth=2,color='mediumslateblue')
#
ax2.set_ylabel(r"Magnetisatie $\bf \mathbb{P} = \frac{\mathbb{M}} {V} \;  (\mu N)$",fontsize=21,weight='bold')
ax2.set_xlabel(r" $\bf x = \frac { \mu \mathbb{B} } { k T}$",fontsize=21,weight='bold')
ax2.set_ylim(0.0,1.0)
ax2.set_xlim(0.,3.)
ax2.locator_params(axis='y',nbins=5) # control the number of y-axis ticks
#%ax2.legend(frameon=False,fontsize=22,ncol=1,loc='best')
fig2.savefig("fig3p2_n.eps")
fig2.savefig("fig3p2_n.pdf")
plt.show(fig2)
```

```
# Figure 3: heat capacity C_{\mathcal{H}} as a function of 1/x
#
fig3 = plt.figure()
fig3.subplots_adjust(bottom=0.2,left=0.2)
ax3  = fig3.add_subplot(111)
# compute the heat capacity
xvalh=np.divide(1.,xm1val)
CH= np.divide(np.power(xvalh,2),np.power(np.cosh(xvalh),2.))
#
#
ax3.plot(xm1val,CH,linestyle='-',linewidth=2,color='mediumslateblue')
#
ax3.set_ylabel(r"Warmtecapaciteit $\bf C_{\mathbb{H}} \; \; (Nk)$ ",fontsize=18,weight='bold')
ax3.set_xlabel(r" $\bf  {x} ^{-1} = \frac { k T} { \mu \mathbb{B} } $",fontsize=18,weight='bold')
ax3.set_ylim(0.0,0.5) ; ax3.set_xlim(0.,6.)
ax3.locator_params(axis='y',nbins=5) # control the number of y-axis ticks
#ax3.legend(frameon=False,fontsize=22,ncol=1,loc='best')
fig3.savefig("fig3p3_n.eps") ; fig3.savefig("fig3p3_n.pdf")
plt.show(fig3)
```

```
# Figure 4: the entropy as a function of 1/x
#
fig4 = plt.figure()
fig4.subplots_adjust(bottom=0.2,left=0.2)
ax4  = fig4.add_subplot(111)
# compute the entropy
#xvalh=np.divide(1.,xm1val)
Entropy= np.add(np.log(2.*np.cosh(xvalh)),-np.multiply(xvalh,np.tanh(xvalh)))
#
#
ax4.plot(xm1val,Entropy,linestyle='-',linewidth=2,color='mediumslateblue')
#
ax4.set_ylabel(r"Entropie $\bf S \;  (Nk)$ ",fontsize=21,weight='bold')
ax4.set_xlabel(r" $ \bf {x} ^{-1} = \frac { k T} { \mu \mathbb{B} } $",fontsize=21,weight='bold')
ax4.set_title(r"$\bf N$ magnetische dipolen in extern magnetisch veld ",color='darkolivegreen',fontsize=18,weight='bold')
ax4.set_ylim(0.0,0.82) ; ax4.set_xlim(0.,6.)
ax4.axhline(y=np.log(2.),color='grey',linestyle='dashed',alpha=0.5) # entropy under conditions of complete disorder
ax4.locator_params(axis='y',nbins=5) # control the number of y-axis ticks
#ax3.legend(frameon=False,fontsize=22,ncol=1,loc='best')
fig4.savefig("fig3p3a_n.eps") ; fig4.savefig("fig3p3a_n.pdf")
plt.show(fig4)
```

#$\color{orange}{\textit{Observables for a paramagnetic material with several spin values}}$


We consider $S=\frac{1}{2}, S=\frac{3}{2}, S=\frac{5}{2}, S=\frac{7}{2} $

The code computes:
- the partition function for one magnetic moment in the field $\vec{\mathcal{B}}$
- the magnetization per degree of freedom (per magnetic moment in the lattice)
- the entropy of the total system
- the susceptibility of the total system

as a function of a dimensionless quantity $x = \frac{\mu \mathcal{B}}{k T}$


```
#----------------------------------------------------------------------
# function that creates a list of m values
#   double physical values:
#    e.g. S_z values with S = 3/2: -3 , -1 , + 1, + 3
def create_list_of_ms(jj):
    mlist=np.linspace(-jj,+jj,jj+1,endpoint=True)
    return mlist
#----------------------------------------------------------------------
```

```
#----------------------------------------------------------------------
# function that computes the partition function, the magnetization,
#     , the susceptibility and the entropy
# x = \mu \mid \vec{B} \mid / k T (dimensionless quantity)
def create_the_obser(xvalues, mlist):
    Z1=np.zeros_like(xvalues)
    M1=np.zeros_like(xvalues)
    M2=np.zeros_like(xvalues)
    magne=np.zeros_like(xvalues)
    suscep=np.zeros_like(xvalues)
    entrop=np.zeros_like(xvalues)
    for m in mlist:
        Z1+=np.exp(xvalues*m)
        M1+=m*np.exp(xvalues*m)
        M2+=m**2*np.exp(xvalues*m)
    magne=np.divide(M1,Z1)
    suscep=np.divide(M2,Z1)-np.power(magne,2)
# S = (E-F) / T
# F = -k T log Z ---> F [ in units \mu \mid \vec{B} \mid] = - k T log Z /x
# E = -magnetization / \mid \vec{B} \mid
# S (in units k) = log Z + x E
    entrop=np.add(np.log(Z1),np.multiply(xvalues,-magne))
    return magne,suscep,entrop
#----------------------------------------------------------------------
```

```
#----------------------------------------------------------------------
# MAIN PROGRAM
# STEP1: create the arrays with  spin projection quantum numbers
listm1=create_list_of_ms(1) # spin 1/2
listm3=create_list_of_ms(3) # spin 3/2
listm5=create_list_of_ms(5) # spin 5/2
listm7=create_list_of_ms(7) # spin 7/2
#
# an array with x values
number_of_xvalues=1000
xvalues=np.linspace(0.,+10.,number_of_xvalues,endpoint=True)
#
# STEP2: create the list of observables (magnetization, susceptibility, entropy)
magne1,suscep1,entrop1=create_the_obser(xvalues,listm1)
magne3,suscep3,entrop3=create_the_obser(xvalues,listm3)
magne5,suscep5,entrop5=create_the_obser(xvalues,listm5)
magne7,suscep7,entrop7=create_the_obser(xvalues,listm7)
```

```
# first figure: the magnetization for a range of spin values
fig=plt.figure(1)
ax=fig.add_subplot(111)
fig.subplots_adjust(bottom=0.2) # create room for x-axis label
#
ax.axis([0.,2.,0.,8.])
#ax.set_title(r'Golfpakketten door $\sum _{i=1}^{i=N}\cos k_i x$',fontsize=20,
#              fontweight='bold')
ax.set_xlabel(r' x=$ \bf (\mu \mid \vec{\mathbb{B}} \mid) / (k T)$', fontsize=20,weight='bold')
ax.set_ylabel(r' magnetisation $\bf \mathbb{M}_z / (N \mu)$', fontsize=20,weight='bold')
ax.plot(xvalues, magne1, 'r-', xvalues, magne3, 'b-', xvalues, magne5, 'g-',xvalues,magne7,'k-')
ax.text(1.5, 1.2,' spin 1/2',color='red',fontsize=18)
ax.text(1.5, 3.2,' spin 3/2',color='blue',fontsize=18)
ax.text(1.5, 5.2,' spin 5/2',color='green',fontsize=18)
ax.text(1.5, 7.2,' spin 7/2',color='black',fontsize=18)
fig.savefig('MagnetizationJ1to7.eps')
plt.show(fig)
```

```
# second figure: susceptibility for a range of spin values
#
fig2=plt.figure(2)
ax2=fig2.add_subplot(111)
fig2.subplots_adjust(bottom=0.2) # create room for x-axis label
#
ax2.axis([0.,2.,0.,25.])
ax2.set_xlabel(r'$\bf x= \frac {\mu \mathbb{B}} {k T}$',fontsize=18,weight='bold')
ax2.set_ylabel(r' susceptibility $\bf  \chi / (N \mu ) / (\mathbb{B}) $', fontsize=18,weight='bold')
ax2.plot(xvalues, suscep1, 'r-', xvalues, suscep3, 'b-', xvalues, suscep5, 'g-',xvalues,suscep7,'k-')
#ax.text(1.5, 1.2,' spin 1/2',color='red',fontsize=18)
#ax.text(1.5, 3.2,' spin 3/2',color='blue',fontsize=18)
#ax.text(1.5, 5.2,' spin 5/2',color='green',fontsize=18)
#ax.text(1.5, 7.2,' spin 7/2',color='black',fontsize=18)
#ax.text(-75.,11.5,r'Golfpakketten door ${\sum _{i=1}^{i=N}\cos k_i x}$',fontsize=22, fontweight='bold')
fig2.savefig('SusceptibilityJ1to7.eps')
plt.show(fig2)
```

```
# third figure with entropy as a function of 1/x for various S values
#
fig3=plt.figure(3)
ax3=fig3.add_subplot(111)
fig3.subplots_adjust(bottom=0.2) # create room for x-axis label
#
ax3.axis([0.,12.,0.,2.2])
ax3.set_xlabel(r'$\bf 1/x=(k T ) / (\mu \mid \vec{\mathbb{B}} \mid)$',fontsize=20,weight='bold')
ax3.set_ylabel(r' entropy $S /(Nk)$ ',fontsize=20,weight='bold')
ax3.plot(1./xvalues, entrop1, 'r-', 1./xvalues, entrop3, 'b--', 1./xvalues,
       entrop5, 'g-',1./xvalues,entrop7,'k--')
ax3.axhline(y=np.log(2.),color='grey',linestyle='dashed',alpha=0.5) # complete disorder for S=1/2
ax3.axhline(y=np.log(4.),color='grey',linestyle='dashed',alpha=0.5) # complete disorder for S=3/2
ax3.axhline(y=np.log(6.),color='grey',linestyle='dashed',alpha=0.5) # complete disorder for S=5/2
ax3.axhline(y=np.log(8.),color='grey',linestyle='dashed',alpha=0.5) # complete disorder for S=7/2
ax3.text(8., 0.5,' spin 1/2',color='red',fontsize=18,fontweight='bold')
ax3.text(8., 1.2,' spin 3/2',color='blue',fontsize=18,fontweight='bold')
ax3.text(8., 1.6,' spin 5/2',color='green',fontsize=18,fontweight='bold')
ax3.text(8., 1.8,' spin 7/2',color='black',fontsize=18,fontweight='bold')
fig3.savefig('EntropyJ1to7.eps')
#
#

plt.show(fig3)
```
