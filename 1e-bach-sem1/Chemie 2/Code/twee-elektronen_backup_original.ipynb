# Twee elektronen in de oneindige potentiaalput 
## De onafhankelijke elektronbenadering

```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import math
import plotly.graph_objects as go
```

```python
# De oplossingen zijn:
a = 1
n = 2
m = 3


def psi_nm(x_1, x_2, length=a, n=n, m=m):
    return (
        (2 / length)
        * np.sin(((n * math.pi) / length) * x_1)
        * np.sin(((m * math.pi) / length) * x_2)
    )


# Visualisatie plot (positie x_1 in functie van x_2)
def get_data(length=a, n=n, m=m):
    x_1 = np.linspace(0, length, 1000)
    x_2 = np.linspace(0, length, 1000)
    X_1, X_2 = np.meshgrid(x_1, x_2)

    # Bereken de golffunctie over het hele rooster
    Z = psi_nm(X_1, X_2, length=a, n=n, m=m)
    return X_1, X_2, Z
```

```python
X_1, X_2, Z = get_data(length=a, n=n, m=m)


def plot_wavefunction(X_1, X_2, Z, n, m):
    plt.close()
    # oppervlakte plot
    plt.contourf(X_1, X_2, Z, levels=20, cmap="coolwarm")
    plt.xlabel(r"$x_1$")
    plt.ylabel(r"$x_2$")
    plt.title(rf"Golffunctie $\Psi_{{n={n},m={m}}}(x_1, x_2)$")
    plt.colorbar()

    plt.show()


plot_wavefunction(X_1, X_2, Z, n, m)
```

```python
# Symmetrische en antisymmetrische golffuncties
def psi_sym(x_1, x_2, length=a, n=n, m=m):
    return (1 / np.sqrt(2)) * (
        psi_nm(x_1, x_2, length=length, n=n, m=m)
        + psi_nm(x_2, x_1, length=length, n=n, m=m)
    )


def psi_antisym(x_1, x_2, length=a, n=n, m=m):
    return (1 / np.sqrt(2)) * (
        psi_nm(x_1, x_2, length=length, n=n, m=m)
        - psi_nm(x_2, x_1, length=length, n=n, m=m)
    )


def get_data_sym_antisym(length=a, n=n, m=m):
    x_1 = np.linspace(0, length, 1000)
    x_2 = np.linspace(0, length, 1000)
    X_1, X_2 = np.meshgrid(x_1, x_2)

    # Bereken de golffunctie over het hele rooster
    Z_sym = psi_sym(X_1, X_2, length=a, n=n, m=m)
    Z_antisym = psi_antisym(X_1, X_2, length=a, n=n, m=m)
    return X_1, X_2, Z_sym, Z_antisym


def plot_wavefunctions_sym_antisym(X_1, X_2, Z_sym, Z_antisym, n, m):
    plt.close()
    # oppervlakte plot
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.contourf(X_1, X_2, Z_sym, levels=20, cmap="coolwarm")
    plt.xlabel(r"$x_1$")
    plt.ylabel(r"$x_2$")
    plt.title(
        rf"Symmetrische Golffunctie $\Psi_{{n={n},m={m}}}^{{\mathrm{{sym}}}}(x_1, x_2)$"
    )
    plt.colorbar()

    plt.subplot(1, 2, 2)
    plt.contourf(X_1, X_2, Z_antisym, levels=20, cmap="coolwarm")
    plt.xlabel(r"$x_1$")
    plt.ylabel(r"$x_2$")
    plt.title(
        rf"Antisymmetrische Golffunctie $\Psi_{{n={n},m={m}}}^{{\mathrm{{antisym}}}}(x_1, x_2)$"
    )
    plt.colorbar()

    plt.tight_layout()
    plt.show()
```

```python
X_1, X_2, Z_sym, Z_antisym = get_data_sym_antisym(length=a, n=n, m=m)
plot_wavefunctions_sym_antisym(X_1, X_2, Z_sym, Z_antisym, n, m)
```

```python
# Hergebruik van de functies uit de vorige stap
a0 = 1.0


def psi_1s(r):
    return (2.0 / np.sqrt(4 * np.pi)) * np.exp(-r / a0)


def psi_2s(r):
    return (
        (1.0 / (2.0 * np.sqrt(2.0) * np.sqrt(4 * np.pi)))
        * (2.0 - (r / a0))
        * np.exp(-r / (2.0 * a0))
    )


def psi_3p_z(r, theta):
    R = (
        (4.0 / (81.0 * np.sqrt(6.0)))
        * (r / a0)
        * (6.0 - (r / a0))
        * np.exp(-r / (3.0 * a0))
    )
    Y = np.sqrt(3.0 / (4.0 * np.pi)) * np.cos(theta)
    return R * Y


# --- 1. 1D Radiale plot (1s en 2s) ---
r_vals = np.linspace(0, 20 * a0, 500)
plt.figure(figsize=(8, 4))
plt.plot(r_vals / a0, psi_1s(r_vals), label="1s")
plt.plot(r_vals / a0, psi_2s(r_vals), label="2s")
plt.axhline(0, color="gray", linestyle="--")
plt.title("Radiale golffuncties")
plt.xlabel("Afstand r / a0")
plt.ylabel("Psi(r)")
plt.legend()
plt.grid(True)
plt.show()

# --- 2. 2D Doorsnede plot (3p_z) in het xz-vlak ---
x = np.linspace(-15 * a0, 15 * a0, 300)
z = np.linspace(-15 * a0, 15 * a0, 300)
X, Z = np.meshgrid(x, z)
R = np.sqrt(X**2 + Z**2)
# Voorkom deling door nul bij r=0
Theta = np.where(R == 0, 0, np.arccos(Z / R))

Psi_3pz = psi_3p_z(R, Theta)

plt.figure(figsize=(6, 6))
plt.contourf(X / a0, Z / a0, Psi_3pz, levels=50, cmap="coolwarm")
plt.colorbar(label="Amplitude")
plt.title("3p_z orbitaal (xz-vlak)")
plt.xlabel("x / a0")
plt.ylabel("z / a0")
plt.axis("equal")
plt.show()
```

```python
# --- 3D Grid genereren ---
n_samples = 1000000
lim = 18 * a0

# Willekeurige punten in een kubus rond de kern
x = np.random.uniform(-lim, lim, n_samples)
y = np.random.uniform(-lim, lim, n_samples)
z = np.random.uniform(-lim, lim, n_samples)

r = np.sqrt(x**2 + y**2 + z**2)
theta = np.where(r == 0, 0, np.arccos(np.clip(z / r, -1.0, 1.0)))

# Bereken kansdichtheid |psi|^2
psi = psi_3p_z(r, theta)
prob = psi**2

# Selecteer alleen punten met een hogere kans (sampling via kansdichtheid)
max_prob = np.max(prob)
mask = np.random.uniform(0, max_prob, n_samples) < prob
x_plot, y_plot, z_plot = x[mask], y[mask], z[mask]
psi_plot = psi[mask]

# --- 3D Plot maken ---
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(projection="3d")

# Scatter plot met kleur op basis van de fase (teken van psi)
sc = ax.scatter(
    x_plot / a0,
    y_plot / a0,
    z_plot / a0,
    c=psi_plot,
    cmap="coolwarm",
    s=1,
    alpha=0.4,
)

ax.set_title("3D 3p_z Orbitaal (Kansdichtheid)")
ax.set_xlabel("x / a0")
ax.set_ylabel("y / a0")
ax.set_zlabel("z / a0")
fig.colorbar(sc, ax=ax, label="Golffunctie waarde (Fase)")
plt.show()
```

```python
def psi_3p_z(x, y, z):
    r = np.sqrt(x**2 + y**2 + z**2)
    # Voorkom deling door nul bij r=0
    theta = np.where(r == 0, 0, np.arccos(np.clip(z / r, -1.0, 1.0)))
    R = (
        (4.0 / (81.0 * np.sqrt(6.0)))
        * (r / a0)
        * (6.0 - (r / a0))
        * np.exp(-r / (3.0 * a0))
    )
    Y = np.sqrt(3.0 / (4.0 * np.pi)) * np.cos(theta)
    return R * Y


# 3D rekenraster (mesh) genereren
def generate_3d_grid(grid_size=60, lim=18 * a0):
    x, y, z = np.mgrid[
        -lim : lim : grid_size * 1j,
        -lim : lim : grid_size * 1j,
        -lim : lim : grid_size * 1j,
    ]
    return x, y, z


x, y, z = generate_3d_grid(grid_size=60, lim=18 * a0)
# Golffunctie evalueren over het volledige volume
psi = psi_3p_z(x, y, z)

# Drempelwaarde bepalen (bijv. 30% van de maximale amplitude om de vorm mooi te capteren)
threshold = np.max(np.abs(psi)) * 0.2
title = "plot"


# De isosurface plot
def plot_isosurface(x=x, y=y, z=z, wavefunction=psi, threshold=threshold, title=title):
    fig = go.Figure(
        data=go.Isosurface(
            x=x.flatten(),
            y=y.flatten(),
            z=z.flatten(),
            value=wavefunction.flatten(),
            isomin=-threshold,
            isomax=threshold,
            surface_count=2,  # Tekent precies twee contouren: één positieve en één negatieve fase
            colorscale="RdBu_r",
            caps=dict(
                x_show=False, y_show=False, z_show=False
            ),  # Zorgt dat de volumegrenzen niet dichtgekleurd worden
        )
    )

    fig.update_layout(
        title=title,
        scene=dict(xaxis_title="x / a0", yaxis_title="y / a0", zaxis_title="z / a0"),
    )
    return fig


plot_isosurface(
    x=x, y=y, z=z, wavefunction=psi, threshold=threshold, title="3p_z Orbitaal"
).show()
```

```python
# sp-hybridisatie: 3D plot van de 2s en 3p_z orbitals samen
def psi_sp_1(x, y, z):
    r = np.sqrt(x**2 + y**2 + z**2)
    psi_2s_val = psi_2s(r)
    psi_3p_z_val = psi_3p_z(x, y, z)
    return (psi_2s_val + psi_3p_z_val) / np.sqrt(2)


def psi_sp_2(x, y, z):
    r = np.sqrt(x**2 + y**2 + z**2)
    psi_2s_val = psi_2s(r)
    psi_3p_z_val = psi_3p_z(x, y, z)
    return (psi_2s_val - psi_3p_z_val) / np.sqrt(2)
```

```python
x, y, z = generate_3d_grid(grid_size=60, lim=18 * a0)
psi_sp1 = psi_sp_1(x, y, z)
psi_sp2 = psi_sp_2(x, y, z)

# Drempelwaarde bepalen (bijv. 30% van de maximale amplitude om de vorm mooi te capteren)
threshold = np.max(np.abs(psi_sp1)) * 0.25

fig = plot_isosurface(
    x=x, y=y, z=z, wavefunction=psi_sp1, threshold=threshold, title=r"sp$^1$ Orbitaal"
)

fig.show()
```

```python
threshold = np.max(np.abs(psi_sp2)) * 0.25
fig = plot_isosurface(
    x=x, y=y, z=z, wavefunction=psi_sp2, threshold=threshold, title=r"sp$^2$ Orbitaal"
)

fig.show()
```
