$$P(x,a)  = 1 + a(x-0.5) $$

```python
import numpy as np
import matplotlib.pyplot as plt
```

$$x \in \{0.89, 0.03, 0.50, 0.36, 0.49\}$$

```python
a = np.linspace(-1, 1, 1000)
x = np.array([0.89, 0.03, 0.50, 0.36, 0.49])


def log_likelihood(a, x):
    return np.sum(np.log(1 + a * (x - 0.5)))
```

```python
plt.plot(a, [log_likelihood(ai, x) for ai in a])
plt.xlabel("a")
plt.ylabel("Log-likelihood")
plt.show()
```

```python
def max_likelihood_estimate(x, a_values=a):
    log_likelihoods = [log_likelihood(ai, x) for ai in a_values]
    max_index = np.argmax(log_likelihoods)
    return a_values[max_index]


max_likelihood_estimate(x)
```
